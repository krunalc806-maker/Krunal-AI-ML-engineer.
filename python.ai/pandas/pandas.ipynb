```python
data={
    "thing":["laptop","bag","cloths","rubber","computer","code"],
    "cost_price":[50,60,55,76,78,80],
    "selling_price":[75,70,95,120,90,100]
}
df=pd.DataFrame(data)
```

```python
df
```

```python
df["profit"]=df["selling_price"]-df["cost_price"]
```

```python
df
```

```python
df["profit_percent"]=df["profit"]/df["cost_price"]*100
```

```python
df
```

```python
print("selling_price",df["selling_price"].mean())
print("selling_price",df["selling_price"].min())
print("selling_price",df["selling_price"].max())
print("selling_price",df["selling_price"].median())
```

```python
df
```

```python
avg_price=df["selling_price"].mean()
df["category"]=df["selling_price"].apply(lambda x:"expensive"
if x>avg_price
                                         else "cheap")
```

df

```python
df["price_normalized"]=(df["selling_price"]-df["selling_price"].min()
)/(
    df["selling_price"].max() - df["selling_price"].min())
```

```python
df
```

```python
total_price=df["selling_price"].sum()
df["sales_price_%"]=(df["selling_price"]/total_price)*100
```

```python
df
```

```python
df.groupby("category")["selling_price"].agg(
    ["count","mean","max","min"]
)
```
